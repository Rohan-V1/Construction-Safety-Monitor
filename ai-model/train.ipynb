```python
from pathlib import Path

LABEL_DIRS = [
    Path("dataset/labels/train"),
    Path("dataset/labels/val"),
    Path("dataset/labels/test"),
]

VALID_CLASSES = {0, 1, 2}

bad = 0
for d in LABEL_DIRS:
    for f in d.glob("*.txt"):
        clean_lines = []
        with open(f) as file:
            for line in file:
                parts = line.strip().split()
                if len(parts) != 5:
                    continue
                cls = int(float(parts[0]))
                x, y, w, h = map(float, parts[1:])
                if cls not in VALID_CLASSES:
                    bad += 1
                    continue
                if not (0 < x < 1 and 0 < y < 1 and 0 < w < 1 and 0 < h < 1):
                    bad += 1
                    continue
                clean_lines.append(line)
        with open(f, "w") as file:
            file.writelines(clean_lines)

print("Bad labels removed:", bad)
```

```python
import cv2
from ultralytics import YOLO

# Load models
person_model = YOLO("yolov8m.pt")
ppe_model = YOLO("runs/detect/train/weights/best.pt")

img = cv2.imread("test1.jpg")
H, W = img.shape[:2]

# Detect persons
person_results = person_model(img, conf=0.5, classes=[0])

for r in person_results:
    for box in r.boxes.xyxy:
        x1, y1, x2, y2 = map(int, box)

        person_crop = img[y1:y2, x1:x2]
        if person_crop.size == 0:
            continue

        # PPE detection inside person
        ppe_results = ppe_model(person_crop, conf=0.4)

        found = {
            "hardhat": False,
            "mask": False,
            "safety_vest": False
        }

        for p in ppe_results:
            for b, cls in zip(p.boxes.xyxy, p.boxes.cls):
                cls_name = ppe_model.names[int(cls)]
                found[cls_name] = True

                px1, py1, px2, py2 = map(int, b)
                cv2.rectangle(
                    img,
                    (x1 + px1, y1 + py1),
                    (x1 + px2, y1 + py2),
                    (0, 255, 0),
                    2
                )
                cv2.putText(
                    img,
                    cls_name,
                    (x1 + px1, y1 + py1 - 5),
                    cv2.FONT_HERSHEY_SIMPLEX,
                    0.5,
                    (0, 255, 0),
                    2
                )

        # Person box
        cv2.rectangle(img, (x1, y1), (x2, y2), (255, 0, 0), 2)

        # Missing PPE logic
        if not found["hardhat"]:
            cv2.putText(img, "NO HARDHAT", (x1, y1 - 10),
                        cv2.FONT_HERSHEY_SIMPLEX, 0.6, (0, 0, 255), 2)

        if not found["safety_vest"]:
            cv2.putText(img, "NO VEST", (x1, y1 - 30),
                        cv2.FONT_HERSHEY_SIMPLEX, 0.6, (0, 0, 255), 2)

# Show result
cv2.imshow("PPE Detection", img)
cv2.waitKey(0)
cv2.destroyAllWindows()
```
